# Recoding Variables from Numeric to Categorical Format

```python

# Mapping Values In Categorical Variables ... Changing them From Numeric to Categories

# INCOME

# Converting  income var to integer
brfss["Income_Cat"] = brfss["Income_Cat"].astype("Int64")

# Ordered income categories
income_order = [
        "Less than $15,000",
        "$15,000 < $25,000",
        "$25,000 < $35,000",
        "$35,000 < $50,000",
        "$50,000 < $100,000",
        "$100,000 < $200,000",
        "$200,000 or more",
        "Missing/Don’t know",
]

# Replacing numeric codes with categories 
brfss["Income_Cat"] = pd.Categorical(
    brfss["Income_Cat"].map({
        1: "Less than $15,000",
        2: "$15,000 < $25,000",
        3: "$25,000 < $35,000",
        4: "$35,000 < $50,000",
        5: "$50,000 < $100,000",
        6: "$100,000 < $200,000",
        7: "$200,000 or more",
        9: np.nan
    }),
    categories = income_order,
    ordered = True
)

# EDUCATION

# Converting  education variable to integer 
brfss["Education_level"] = brfss["Education_level"].astype("Int64")

# Ordered education categories
education_order = [
    "Did not graduate High School",
    "Graduated High School",
    "Attended College or Technical School",
    "Graduated from College or Technical School",
    "Don’t know/Not sure/Missing"
]

# Replace numeric codes with categories and create an ordered categorical
brfss["Education_level"] = pd.Categorical(
    brfss["Education_level"].map({
        1: "Did not graduate High School",
        2: "Graduated High School",
        3: "Attended College or Technical School",
        4: "Graduated from College or Technical School",
        9: np.nan
    }),
    categories=education_order,
    ordered=True
)

# ALCOHOL

# Converting the column to integer
brfss["Heavy_Drinker"] = brfss["Heavy_Drinker"].astype("Int64")

# Map numeric codes 
brfss["Heavy_Drinker"] = pd.Categorical(
    brfss["Heavy_Drinker"].map({
        1: "No",
        2: "Yes",
        9: np.nan
    })
)

# SMOKING 

# Converting  column to integer 
brfss["Smoker"] = brfss["Smoker"].astype("Int64")

# Map numeric codes to categor
brfss["Smoker"] = pd.Categorical(
    brfss["Smoker"].map({
        1: "No",
        2: "Yes",
        9: np.nan 
    })
)

# OBESE

# Converting column to integer 
brfss["Obese"] = brfss["Obese"].astype("Int64")

# Mapping numeric codes to categories
brfss["Obese"] = brfss["Obese"].map({
    1: "No",
    2: "Yes",
    9: np.nan 
})

# SEX

# Converting to integer 
brfss["Sex"] = brfss["Sex"].astype("Int64")
#brfss['Sex'] = brfss['Sex'].map({'Male': 1, 'Female': 2})

#Mapping numeric codes to categories
brfss["Sex"] = brfss["Sex"].map ({1: "Male",2: "Female"})

# AGE GROUP

# Converting column to integer 
brfss["Age_group"] = brfss["Age_group"].astype("Int64")

# Defining the ordered categories
age_order = [
    "Age 18 to 24",
    "Age 25 to 34",
    "Age 35 to 44",
    "Age 45 to 54",
    "Age 55 to 64",
    "Age 65 or older"
]

# Map numeric codes to categories 
brfss["Age_group"] = pd.Categorical(
    brfss["Age_group"].map({
        1: "Age 18 to 24",
        2: "Age 25 to 34",
        3: "Age 35 to 44",
        4: "Age 45 to 54",
        5: "Age 55 to 64",
        6: "Age 65 or older"
    }),
    categories=age_order,
    ordered=True
)

# PHYSICAL ACTIVITY

# Converting column to integer 
brfss["Physical_Activity"] = brfss["Physical_Activity"].astype("Int64")

brfss["Physical_Activity"] = pd.Categorical(
    brfss["Physical_Activity"].map({
        1: "Had physical activity or exercise",
        2: "No physical activity or exercise in last 30 days",
        9: np.nan
    })
)

# ASTHMA

brfss["Asthma"] = pd.Categorical(
    brfss["Asthma"].map({
        1: "Yes",
        2: "No",
        7: np.nan,
        9: np.nan
    })
)

# DIABETES

brfss["Diabetes"] = pd.Categorical(
    brfss["Diabetes"].map({
        1: "Yes",
        2: "No",
        3: "No",
        4: "No",
        7: np.nan,
        9: np.nan
    })
)

# KIDNEY DISEASE

brfss["Kidney_Disease"] = pd.Categorical(
    brfss["Kidney_Disease"].map({
        1: "Yes",
        2: "No",
        7: np.nan,
        9: np.nan
    })
)

# ARTHRITIS

brfss["Arthritis"] = pd.Categorical(
    brfss["Arthritis"].map({
        1: "Yes",
        2: "No"
    })
)

# HIGH CHOLESTEROL

brfss["High_Cholesterol"] = pd.Categorical(
    brfss["High_Cholesterol"].map({
        1: "Yes",
        2: "No",
        7: np.nan,
        9: np.nan,
        999: np.nan  # Years where the variable wasn't captured (2022,2024)
    })
)


# HYPERTENSION

brfss["High_BP"] = pd.Categorical(
    brfss["High_BP"].map({
        1: "Yes",
        2: "No",
        7: np.nan,
        9: np.nan,
        999: np.nan  # Years where the variable wasn't captured (2022,2024)
    })
)
```

# Data Preprocessing and Feature Engineering

```python
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OrdinalEncoder, LabelEncoder, StandardScaler
from sklearn.impute import SimpleImputer, KNNImputer
from sklearn.preprocessing import LabelEncoder
    

#  Ordinal 
ordinal_cols = ["Age_group", "Education_level", "Income_Cat"]
ordinal_orders = [
    ["Age 18 to 24", "Age 25 to 34", "Age 35 to 44",
     "Age 45 to 54", "Age 55 to 64", "Age 65 or older"],
    ["Did not graduate High School", "Graduated High School",
     "Attended College or Technical School",
     "Graduated from College or Technical School",
     "Don't know/Not sure/Missing"],
    ["Less than $15,000", "$15,000 < $25,000", "$25,000 < $35,000",
     "$35,000 < $50,000", "$50,000 < $100,000",
     "$100,000 < $200,000", "$200,000 or more", "Missing/Don't know"],
]

# Ordinal pipeline with mean imputation 
ordinal_pipeline = Pipeline(steps=[
    ("impute", SimpleImputer(strategy="most_frequent")),
    ("encode", OrdinalEncoder(
                   categories=ordinal_orders,
                   handle_unknown="use_encoded_value",
                   unknown_value=np.nan
               )),
])


#  Binary 
binary_cols = ["Sex", "Smoker", "Heavy_Drinker", "Physical_Activity","Obese", "Asthma","Kidney_Disease","Arthritis", "High_Cholesterol", "High_BP" ]

binary_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='most_frequent')),  
    ('encoder', OrdinalEncoder())                         
])

# Numeric
numeric_cols = ["BMI"]

numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),   
    ('scaler', StandardScaler())                   
])

# Preprocessor
preprocessor = ColumnTransformer(
    transformers=[
    ('ordinal',  ordinal_pipeline,  ordinal_cols),
    ('binary',   binary_pipeline,   binary_cols),
    ('numeric',  numeric_pipeline,  numeric_cols),
], remainder='drop')

#  Fit & Transform
X = brfss[ordinal_cols + binary_cols + numeric_cols]
y = brfss['Diabetes']


X_processed = preprocessor.fit_transform(X)

# Converting Back to a Dataframe
feature_names = ordinal_cols + binary_cols + numeric_cols
X_processed_data = pd.DataFrame(X_processed, columns=feature_names)

# dropping Missing values in the target variable , 43037 out of 457669 observations
mask = y.notna()
y = y[mask]

# Encoding Y
le = LabelEncoder()
y = le.fit_transform(y)

X_processed_data = X_processed_data[mask]
```

```python
# Summary stats
print("Shape:", X_processed_data.shape)
print("\nMissing values:\n", X_processed_data.isnull().sum())
print("\nSample:\n", X_processed_data.head())
```

# Modelling

```python
from sklearn.model_selection import cross_val_score
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from xgboost import XGBClassifier
from sklearn.metrics import classification_report, roc_auc_score
from imblearn.pipeline import Pipeline
from imblearn.over_sampling import SMOTE

def run_classifier(classifier):
    X_train, X_test, y_train, y_test = train_test_split(
        X_processed_data, y,
        test_size=0.2,
        random_state=42,
        stratify=y
    )

    # Applying SMOTE since the target variable is imbalanced
    pipeline = Pipeline([
        ("smote",      SMOTE(random_state=42)),
        ("classifier", classifier)
    ])

    # 5-fold CV
    cv_scores = cross_val_score(pipeline, X_train, y_train, scoring="f1_macro", cv=5)
    print(f"\n{classifier.__class__.__name__}")
    print(f"CV F1 scores : {cv_scores.round(4)}")
    print(f"Mean F1      : {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")

    # Model Evaluation
    pipeline.fit(X_train, y_train)
    y_pred  = pipeline.predict(X_test)
    y_proba = pipeline.predict_proba(X_test)[:, 1]

    print(f"\n{classification_report(y_test, y_pred, target_names=['Not Diabetes', 'Diabetes'])}")
    print(f"ROC-AUC: {roc_auc_score(y_test, y_proba):.4f}")

    return cv_scores, pipeline
```

# Logistic Regression With BMI and Obese

```python
lr_scores, lr_pipeline = run_classifier(LogisticRegression(max_iter=1000, random_state=42))

# Logistic Regression Coefficients and Odds Ratio

fitted_lr = lr_pipeline.named_steps["classifier"]

coef_df = pd.DataFrame({
    "Feature":     X_processed_data.columns.tolist(),
    "Coefficient": fitted_lr.coef_[0],
    "Odds_Ratio":  np.exp(fitted_lr.coef_[0])
}).sort_values("Coefficient", ascending=False)

print(coef_df)
```

# Random Forest and XGBoost with Bayesian Tuning

```python
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold
from sklearn.ensemble import RandomForestClassifier

# ── Sample the data to fit in memory ──────────────────────────────────────
# Stratified sample of 20% — preserves class distribution
sample_idx = pd.Series(range(len(X_processed_data))).sample(
    frac=0.2, random_state=42
)
X_sample = X_processed_data.iloc[sample_idx].reset_index(drop=True)
y_sample = y[sample_idx.values]

print(f"Full dataset : {X_processed_data.shape}")
print(f"Sample size  : {X_sample.shape}")
print(f"Class distribution: {pd.Series(y_sample).value_counts().to_dict()}")

# ── Train/test split on sample ────────────────────────────────────────────
X_train_rf, X_test_rf, y_train_rf, y_test_rf = train_test_split(
    X_sample, y_sample,
    test_size    = 0.2,
    random_state = 42,
    stratify     = y_sample
)

# ── Model ─────────────────────────────────────────────────────────────────
cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

rf_pipe = Pipeline([
    ("classifier", RandomForestClassifier(
        random_state = 42,
        class_weight = "balanced",
        n_jobs       = 1
    ))
])

rf_params = {
    "classifier__n_estimators":      [100, 200],
    "classifier__max_depth":         [5, 10, 20],
    "classifier__min_samples_split": [2, 5],
    "classifier__min_samples_leaf":  [1, 2],
    "classifier__max_features":      ["sqrt", "log2"],
}

rf_search = RandomizedSearchCV(
    estimator           = rf_pipe,
    param_distributions = rf_params,
    scoring             = "f1_macro",
    cv                  = cv,
    n_iter              = 5,
    n_jobs              = 1,
    random_state        = 42,
    verbose             = 1,
    refit               = True
)

rf_search.fit(X_train_rf, y_train_rf)

print(f"\nBest RF F1-macro : {rf_search.best_score_:.4f}")
print("Best RF params   :", rf_search.best_params_)

# ── Evaluate ──────────────────────────────────────────────────────────────
best_rf    = rf_search.best_estimator_
y_pred_rf  = best_rf.predict(X_test_rf)
y_proba_rf = best_rf.predict_proba(X_test_rf)[:, 1]

print(f"\n{classification_report(y_test_rf, y_pred_rf, target_names=['Not Diabetes', 'Diabetes'])}")
print(f"ROC-AUC: {roc_auc_score(y_test_rf, y_proba_rf):.4f}")

# ── Feature Importance ────────────────────────────────────────────────────
fitted_rf = best_rf.named_steps["classifier"]
rf_importance = pd.DataFrame({
    "Feature":    X_sample.columns.tolist(),
    "Importance": fitted_rf.feature_importances_
}).sort_values("Importance", ascending=False)

print("\nFeature Importances:")
print(rf_importance)

plt.figure(figsize=(10, 6))
plt.barh(rf_importance["Feature"], rf_importance["Importance"])
plt.xlabel("Importance Score")
plt.title("RF Feature Importance (Tuned — 20% Sample)")
plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

X_train_xgb, X_test_xgb, y_train_xgb, y_test_xgb = train_test_split(
    X_processed_data, y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

# Use scale_pos_weight instead of SMOTE — handles imbalance without memory overhead
neg = (y_train_xgb == 0).sum()
pos = (y_train_xgb == 1).sum()
scale = neg / pos
print(f"scale_pos_weight: {scale:.2f}  (neg={neg}, pos={pos})")

xgb_pipe = Pipeline([
    ("classifier", XGBClassifier(
        eval_metric    = "logloss",
        random_state   = 42,
        scale_pos_weight = scale,   # replaces SMOTE
        tree_method    = "hist",    # faster and more memory efficient
        device         = "cpu",
        n_jobs         = 1          # avoid parallel memory spikes
    ))
])

# Reduced parameter grid to avoid memory issues
xgb_params = {
    "classifier__n_estimators":     [100, 200, 300],
    "classifier__max_depth":        [3, 5],
    "classifier__learning_rate":    [0.05, 0.1, 0.2],
    "classifier__subsample":        [0.6, 0.8],
    "classifier__colsample_bytree": [0.6, 0.8],
    "classifier__gamma":            [0, 1, 3],
    "classifier__reg_alpha":        [0, 0.1, 1.0],
    "classifier__reg_lambda":       [1, 2, 5],
}

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

xgb_search = RandomizedSearchCV(
    estimator          = xgb_pipe,
    param_distributions = xgb_params,
    scoring            = "f1_macro",
    cv                 = cv,
    n_iter             = 10,
    n_jobs             = 1,          # 1 job to avoid memory spikes
    random_state       = 42,
    verbose            = 1,
    refit              = True
)

xgb_search.fit(X_train_xgb, y_train_xgb)

print(f"\nBest XGB F1-macro : {xgb_search.best_score_:.4f}")
print("Best XGB params   :", xgb_search.best_params_)

# Evaluate on test set
best_xgb    = xgb_search.best_estimator_
y_pred_xgb  = best_xgb.predict(X_test_xgb)
y_proba_xgb = best_xgb.predict_proba(X_test_xgb)[:, 1]

print(f"\n{classification_report(y_test_xgb, y_pred_xgb, target_names=['Not Diabetes', 'Diabetes'])}")
print(f"ROC-AUC: {roc_auc_score(y_test_xgb, y_proba_xgb):.4f}")
```
